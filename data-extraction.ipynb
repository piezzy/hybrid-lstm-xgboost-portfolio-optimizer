```python
import yfinance as yf
import pandas as pd

LQ45_TICKERS = [
    'ADRO.JK','AKRA.JK','AMMN.JK','AMRT.JK','ANTM.JK',
    'ASII.JK','BBCA.JK','BBNI.JK','BBRI.JK','BMRI.JK',
    'BRPT.JK','CPIN.JK','CTRA.JK','EXCL.JK','GOTO.JK',
    'ICBP.JK','INCO.JK','INDF.JK','INTP.JK','ITMG.JK',
    'JPFA.JK','KLBF.JK','MDKA.JK','MEDC.JK','PGAS.JK',
    'PTBA.JK','SIDO.JK','SMGR.JK','TLKM.JK','UNTR.JK',
    'UNVR.JK','ESSA.JK','BBYB.JK','MIKA.JK','TOWR.JK',
    'HRUM.JK','INKP.JK','TKIM.JK','BUKA.JK','PWON.JK',
    'MAPA.JK','ERAA.JK','ELSA.JK','SMRA.JK','WSKT.JK'
]

START = "2017-01-01"
END   = "2025-11-30"

print("Downloading all tickers into one big dataframe...")

data = yf.download(
    LQ45_TICKERS,
    start=START,
    end=END,
    group_by='column',
    auto_adjust=False
)

data.to_csv("all_data.csv")
print(" Saved: all_data.csv")

print("\nDownloading IHSG Benchmark (^JKSE)...")
ihsg = yf.download(
    "^JKSE",
    start=START,
    end=END,
    auto_adjust=False,
    progress=False
)

ihsg.to_csv("IHSG.csv")
print(f"Saved: IHSG.csv (Shape: {ihsg.shape})")

print("\nSelesai.")
```
